##  Ajout de la cible Soil_Moisture_J1

Dans ce notebook, nous ajoutons une nouvelle colonne appelée `Soil_Moisture_J1`.

Cette colonne représente une estimation de l'humidité du sol au jour suivant.



```python
import pandas as pd 
import numpy as np 
```

```python

df = pd.read_csv(r"C:\Users\kerro\OneDrive\Bureau\smart_agri2\data\irrigation_prediction_11cols.csv")
df.head()
```

```python
df["Soil_Moisture_J1"] = (
    df["Soil_Moisture"]
    + 0.015 * df["Rainfall_mm"]
    + 0.05 * df["Humidity"]
    - 0.30 * df["Temperature_C"]
    - 0.40 * df["Wind_Speed_kmh"]
)

df["Soil_Moisture_J1"] = df["Soil_Moisture_J1"].clip(0, 100)



```

```python
df[[ "Soil_Moisture","Soil_Moisture_J1"]].head(15)
```

```python
df.to_csv("../data/irrigation_prediction_12cols.csv", index=False)
```
